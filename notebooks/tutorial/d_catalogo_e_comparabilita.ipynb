# Catalogo degli indicatori e comparabilita'

Questo notebook mostra come usare il catalogo degli indicatori censuari e la matrice di comparabilita' tra anni. Non richiede download live perche' il catalogo e' distribuito insieme al pacchetto.


```python
from istat_census_data import (
    get_census_index_catalog,
    get_comparable_indexes,
    list_census_indexes,
)
```

## Elencare i codici sorgente di un anno


```python
codici_2021 = list_census_indexes(year=2021)
codici_2021[:10]
```

`list_census_indexes()` restituisce i codici ISTAT nello stesso ordine del tracciato. Usa questi codici quando vuoi selezionare variabili specifiche dai layer finali del censimento.


## Esaminare i metadati del catalogo


```python
catalogo_2021 = get_census_index_catalog(year=2021)
catalogo_2021.head()
```

Le colonne piu' utili sono `source_code`, `label_it`, `theme`, `measure_type`, `is_additive` e `comparable_group`. Per workflow H3 area-weighted devi scegliere conteggi additivi, non identificativi o campi testuali.


## Confrontare indicatori tra anni


```python
comparabili_strict = get_comparable_indexes(
    years=[1991, 2001, 2011, 2021],
    strict=True,
)
comparabili_strict[["comparable_group", "year", "source_code", "label_it"]].head()
```

La comparabilita' strict restituisce solo i gruppi presenti in tutti gli anni richiesti. Usa `strict=False` quando vuoi includere anche gruppi disponibili solo in una parte degli anni e leggere le colonne `available_years` e `missing_years`.


```python
comparabili_parziali = get_comparable_indexes(
    years=[1991, 2001, 2011, 2021],
    strict=False,
)
comparabili_parziali[["comparable_group", "available_years", "missing_years"]].drop_duplicates().head()
```
