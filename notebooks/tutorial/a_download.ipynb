# Download dei dati censuari grezzi

Questo notebook mostra la prima fase del workflow: scaricare dati censuari, geodati e confini amministrativi con `download_census()`.

La cella di download e' disattivata di default per evitare download accidentali. Prima di abilitarla scegli una cartella di lavoro, verifica lo spazio disponibile e imposta una lista di regioni limitata se stai facendo una prova.


## Prerequisiti

Installa il pacchetto con `pip install istat-census-data` oppure lavora dall'ambiente Poetry del repository. Il notebook usa l'import supportato da `2.0.0` in poi: `istat_census_data`.

Il download usa fonti ISTAT remote, quindi richiede rete e puo' richiedere tempo. Gli esempi usano la Campania (`region_list=[15]`) per mantenere la prova piu' piccola di un download nazionale.


## Import delle librerie


```python
from pathlib import Path

from istat_census_data.executor.download import download_census
```

## Parametri di lavoro

`output_path` e' la cartella radice in cui verranno salvati gli archivi scaricati ed estratti. Cambiala prima di eseguire il notebook se vuoi separare i dati tutorial dai dati di produzione.


```python
output_path = Path("census_tutorial")
years = [1991, 2001, 2011, 2021]
region_list = [15]
```

## Esecuzione del download

Imposta `RUN_LIVE_DOWNLOAD = True` solo quando vuoi eseguire davvero il download. Lasciandolo a `False`, il notebook resta sicuro da eseguire durante una lettura o una revisione.


```python
RUN_LIVE_DOWNLOAD = False

if RUN_LIVE_DOWNLOAD:
    download_census(
        years=years,
        region_list=region_list,
        output_data_folder=output_path,
    )
else:
    print("Download skipped. Set RUN_LIVE_DOWNLOAD = True to download ISTAT data.")
```

## Output atteso e passo successivo

Dopo il download la cartella `census_tutorial` contiene i dati grezzi necessari al preprocessamento. Il passo successivo e' il notebook `b_preprocessamento.ipynb`, che legge la cartella di preprocessing creata sotto lo stesso percorso.
