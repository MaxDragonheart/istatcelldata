# Finalize Census Data into a GeoPackage

This notebook creates the final `census_data.gpkg` file from preprocessed data. The final GeoPackage has one census layer per selected year and is the normal input for GIS use and H3 estimation.


```python
from pathlib import Path

from istat_census_data.executor.process import finalize_census_data
```

## Configure Finalization


```python
census_data_path = Path("census_tutorial")
years = [1991, 2001, 2011, 2021]
delete_preprocessed_data = True
```

Set `delete_preprocessed_data=False` when you want to keep the intermediate GeoPackage for inspection or troubleshooting.


```python
RUN_FINALIZATION = False

if RUN_FINALIZATION:
    finalize_census_data(
        census_data_path=census_data_path,
        years=years,
        delete_preprocessed_data=delete_preprocessed_data,
    )
else:
    print("Finalization skipped. Set RUN_FINALIZATION = True after preprocessing data.")
```

## Expected Output

A successful run writes `census_tutorial/census_data.gpkg`. Typical layers are `census1991`, `census2001`, `census2011`, and `census2021`.
