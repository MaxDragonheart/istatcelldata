# Census Indicator Catalogue and Comparability

This notebook uses the packaged census indicator catalogue introduced after the `2.0.0` import migration. It does not require live downloads because the catalogue is shipped with the package.


```python
from istat_census_data import (
    get_census_index_catalog,
    get_comparable_indexes,
    list_census_indexes,
)
```

## List Source Codes for One Year


```python
available_2021_codes = list_census_indexes(year=2021)
available_2021_codes[:10]
```

`list_census_indexes()` returns ISTAT source field codes in trace order. Use these codes when you need to select specific variables from finalized census layers.


## Inspect Catalogue Metadata


```python
catalogue_2021 = get_census_index_catalog(year=2021)
catalogue_2021.head()
```

Important fields include `source_code`, `label_it`, `theme`, `measure_type`, `is_additive`, and `comparable_group`. H3 area-weighted workflows should use additive counts, not identifiers or descriptive text fields.


## Compare Indicators Across Years


```python
strict_comparable = get_comparable_indexes(
    years=[1991, 2001, 2011, 2021],
    strict=True,
)
strict_comparable[["comparable_group", "year", "source_code", "label_it"]].head()
```

Strict comparability returns groups present in every selected year. Use `strict=False` when you also want metadata about groups that are missing in some years.


```python
partial_comparable = get_comparable_indexes(
    years=[1991, 2001, 2011, 2021],
    strict=False,
)
partial_comparable[["comparable_group", "available_years", "missing_years"]].drop_duplicates().head()
```
