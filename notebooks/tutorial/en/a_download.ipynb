# Download Raw Census Data

This notebook introduces the first package workflow step: downloading ISTAT census data, geodata, and administrative boundaries with `download_census()`.

The live download cell is disabled by default. Enable it only after choosing a local data folder and confirming that a remote download is acceptable for your session.


## Prerequisites

Install the package with `pip install istat-census-data`, or run the notebook from the repository Poetry environment. New code should import from `istat_census_data`.

The example limits geodata downloads to Campania with `region_list=[15]`. Use an empty list only when you really want all regions.


```python
from pathlib import Path

from istat_census_data.executor.download import download_census
```

## Choose the Download Scope


```python
output_path = Path("census_tutorial")
years = [1991, 2001, 2011, 2021]
region_list = [15]
```

## Run the Download

Set `RUN_LIVE_DOWNLOAD = True` to execute the remote download. Keeping the flag set to `False` makes the notebook safe to run during documentation checks.


```python
RUN_LIVE_DOWNLOAD = False

if RUN_LIVE_DOWNLOAD:
    download_census(
        years=years,
        region_list=region_list,
        output_data_folder=output_path,
    )
else:
    print("Download skipped. Set RUN_LIVE_DOWNLOAD = True to download ISTAT data.")
```

## Expected Output

After a successful download, `census_tutorial` contains the raw files needed by the preprocessing notebook. Keep the folder structure intact; the next workflow step expects it.
