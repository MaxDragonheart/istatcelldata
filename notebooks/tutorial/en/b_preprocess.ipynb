# Preprocess Raw Census Data

This notebook shows how to convert downloaded raw data into the intermediate GeoPackage used by the finalization step.

Run it after the download notebook. The operational cell is disabled by default because it requires local downloaded files.


```python
from pathlib import Path

from istat_census_data.executor.preprocess import preprocess_census
```

## Configure the Input Folder and Filters


```python
processed_data_folder = Path("census_tutorial/preprocessing")
years = [1991, 2001, 2011, 2021]
municipalities_code: list[int] = []
delete_download_folder = True
```

Use `municipalities_code` to keep only selected ISTAT municipality codes. Use `delete_download_folder=False` when you want to inspect or archive the raw files after preprocessing.


```python
RUN_PREPROCESSING = False

if RUN_PREPROCESSING:
    preprocess_census(
        processed_data_folder=processed_data_folder,
        years=years,
        delete_download_folder=delete_download_folder,
        municipalities_code=municipalities_code,
    )
else:
    print("Preprocessing skipped. Set RUN_PREPROCESSING = True after downloading data.")
```

## Expected Output

The preprocessing step writes an intermediate GeoPackage below `census_tutorial/preprocessing`. The finalization notebook reads that intermediate output and creates `census_data.gpkg`.
