# Processamento finale

Questo notebook mostra come creare il GeoPackage finale con `finalize_census_data()` dopo il preprocessamento.

Il risultato finale e' il file usato dalle analisi GIS e dalle procedure H3: `census_data.gpkg`.


## Prerequisiti

La cartella `census_tutorial/preprocessing` deve contenere il GeoPackage intermedio creato dal notebook di preprocessamento. La cella operativa e' disattivata di default per evitare errori quando i dati intermedi non sono presenti.


## Import delle librerie


```python
from pathlib import Path

from istat_census_data.executor.process import finalize_census_data
```

## Parametri di lavoro

`census_data_path` e' la cartella radice in cui verra' scritto `census_data.gpkg`. Usa `delete_preprocessed_data=False` se vuoi conservare il GeoPackage intermedio per controlli o debug.


```python
census_data_path = Path("census_tutorial")
years = [1991, 2001, 2011, 2021]
delete_preprocessed_data = True
```

## Esecuzione della finalizzazione


```python
RUN_FINALIZATION = False

if RUN_FINALIZATION:
    finalize_census_data(
        census_data_path=census_data_path,
        years=years,
        delete_preprocessed_data=delete_preprocessed_data,
    )
else:
    print("Finalization skipped. Set RUN_FINALIZATION = True after preprocessing data.")
```

## Output atteso

Al termine avrai `census_tutorial/census_data.gpkg`, con layer come `census1991`, `census2001`, `census2011` e `census2021`. Questo file e' l'input piu' comodo per QGIS, GeoPandas e per le procedure H3.
