# Preprocessamento dei dati censuari grezzi

Questo notebook mostra come trasformare i dati scaricati in un GeoPackage intermedio con `preprocess_census()`.

Eseguilo dopo il download. La cella operativa e' disattivata di default per evitare errori se i dati grezzi non sono presenti.


## Prerequisiti

La cartella `census_tutorial` deve contenere i dati scaricati dal notebook precedente. Il preprocessamento puo' filtrare i comuni con `municipalities_code` e puo' eliminare i file di download al termine del processo.


## Import delle librerie


```python
from pathlib import Path

from istat_census_data.executor.preprocess import preprocess_census
```

## Parametri di lavoro

`processed_data_folder` punta alla sottocartella di preprocessing. Il parametro `delete_download_folder` controlla la pulizia dei file scaricati dopo la creazione del GeoPackage intermedio.


```python
processed_data_folder = Path("census_tutorial/preprocessing")
years = [1991, 2001, 2011, 2021]
municipalities_code: list[int] = []
delete_download_folder = True
```

## Esecuzione del preprocessamento


```python
RUN_PREPROCESSING = False

if RUN_PREPROCESSING:
    preprocess_census(
        processed_data_folder=processed_data_folder,
        years=years,
        delete_download_folder=delete_download_folder,
        municipalities_code=municipalities_code,
    )
else:
    print("Preprocessing skipped. Set RUN_PREPROCESSING = True after downloading data.")
```

## Output atteso e passo successivo

Il risultato principale e' un GeoPackage intermedio nella cartella di preprocessing. Il notebook `c_processamento.ipynb` usa questi dati per creare il GeoPackage finale `census_data.gpkg` con un layer per anno censuario.
